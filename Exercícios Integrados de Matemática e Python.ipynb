📘 Exercícios Integrados de Matemática e Python

####1 – Operações Básicas e Variáveis

```python
try:
    # Recebendo os dois números inteiros do usuário
    numero1 = int(input("Digite o primeiro número inteiro: "))
    numero2 = int(input("Digite o segundo número inteiro: "))

    # Realizando as operações
    soma = numero1 + numero2
    subtracao = numero1 - numero2
    multiplicacao = numero1 * numero2

    # Exibindo os resultados formatados
    print("\n=== Resultados ===")
    print(f"{numero1} + {numero2} = {soma}")
    print(f"{numero1} - {numero2} = {subtracao}")
    print(f"{numero1} * {numero2} = {multiplicacao}")
    
    # Tratando a divisão 
    if numero2 != 0:
        divisao = numero1 / numero2
        print(f"{numero1} / {numero2} = {divisao:.2f}") 
        
        # Formata para 2 casas decimais
    else:
        print("Divisão: Não é possível dividir por zero.")

except ValueError:
    print("Erro: Por favor, digite apenas números inteiros.")
except Exception as e:
    print(f"Ocorreu um erro inesperado: {e}")
```

####4 – Condições e Cálculos

```python
# Recebendo os três números reais
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
num3 = float(input("Digite o terceiro número: "))

# Encontrando o maior número
if num1 >= num2 and num1 >= num3:
    maior = num1
elif num2 >= num1 and num2 >= num3:
    maior = num2
else:
    maior = num3

# Encontrando o menor número
if num1 <= num2 and num1 <= num3:
    menor = num1
elif num2 <= num1 and num2 <= num3:
    menor = num2
else:
    menor = num3

# Calculando a média aritmética
media = (num1 + num2 + num3) / 3

# Exibindo os resultados
print("Maior número:", maior)
print("Menor número:", menor)
print("Média aritmética:", media)
```

####5 – Funções, Listas e Estatísticas

```python
# Recebendo uma lista de números digitados pelo usuário
entrada = input("Digite os números separados por espaço: ")
lista = list(map(float, entrada.split()))

# Calcula a média
media = sum(lista) / len(lista)

# Encontra o maior e o menor valor
maior = max(lista)
menor = min(lista)

# Conta quantos números são maiores que a média
maiores_que_media = sum(1 for x in lista if x > media)

# Exibe os resultados
print(f"Média: {media}")
print(f"Maior valor: {maior}")
print(f"Menor valor: {menor}")
print(f"Quantidade de números maiores que a média: {maiores_que_media}")
```

📘 Exercícios de Funções Matemáticas e Python

####1 – Funções Simples

```python

# Solicita ao usuário que digite um valor para x
x = float(input("Digite o valor de x: "))

# Calcula f(x)
f = 3 * x + 2

# Exibe o resultado
print("O valor de f(x) = 3x + 2 é:", f)
```

####2 – Funções Lineares em Python

```python
def salario(h):
    return 20 * h + 200

# Exemplo de uso
print(salario(10))  # saída: 400
```

####3 – Funções Quadráticas

```python
# Definindo a função
def f(x):
    return x**2 - 3*x + 2

# Calculando e exibindo os valores de f(x) para x de -5 até 5
for x in range(-5, 6):  # range de -5 até 5 (inclusive)
    resultado = f(x)
    print(f"f({x}) = {resultado}")
```

####4 – Composição de Funções

```python
def f(x):
    return x**2 + 2

def g(x):
    return x - 4

print(f(g(6)))
```

####5 – Função de Lucro

```python
# Implementação da função
def eu(t):
    return 12 * t - (6 * t + 150)

# Mostrando os resultados de t = 0 até t = 20
for t in range(21):
    resultado = eu(t)
    print(f"t = {t}, eu(t) = {resultado}")
```
