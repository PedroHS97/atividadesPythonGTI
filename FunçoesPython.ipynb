```python
import statistics as st
def notas1 ():
    n1 = float(input("Digite a N1"))
    n2 = float(input("Digite a N2"))
    n3 = float(input("Digite a N3"))

    valor = [n1,n2,n3]

    media = round(st.mean(valor),1)
    print("sua media e " , media)
```

```python
notas1()
```

```python
import statistics as st
def notas2 ():
    n1 = float(input("Digite a N1"))
    n2 = float(input("Digite a N2"))
    n3 = float(input("Digite a N3"))

    valor = [n1,n2,n3]

    media = round(st.mean(valor),1)
    return media
```
